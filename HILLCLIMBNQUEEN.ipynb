```python
import random

def get_h_score(state):
    """Calculate the number of conflicting pairs of queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict
            if state[i] == state[j]:
                conflicts += 1
            # Diagonal conflict
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Generate neighbors and find the one with the lowest heuristic score."""
    n = len(state)
    current_h = get_h_score(state)
    best_state = list(state)
    best_h = current_h

    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_h = get_h_score(neighbor)

                if neighbor_h < best_h:
                    best_h = neighbor_h
                    best_state = neighbor

    return best_state, best_h

def hill_climbing_n_queens(n=8, max_restarts=100):
    """Solve N-Queens using Hill Climbing with random restarts."""
    for restart in range(max_restarts):
        # Initialize random state (each index represents a column, value represents row position)
        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_h = get_h_score(current_state)

        while True:
            neighbor, neighbor_h = get_best_neighbor(current_state)

            if neighbor_h >= current_h:
                # Stuck in local minimum or reached global minimum
                break
            current_state = neighbor
            current_h = neighbor_h

        if current_h == 0:
            print(f"Solution found after {restart} restarts:")
            print_board(current_state)
            return current_state

    print("Failed to find a solution within the limit of restarts.")
    return None

def print_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += "Q "
            else:
                line += ". "
        print(line)
    print(f"State representation (rows for each col): {state}\n")

# Run the algorithm for N=8
solution = hill_climbing_n_queens(8)
```

```python
import random

def get_detailed_h_score(state):
    """Calculate the number of conflicting pairs of queens (h value)."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict
            if state[i] == state[j]:
                conflicts += 1
            # Diagonal conflict
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Find the best neighbor state with the lowest conflict score."""
    n = len(state)
    current_h = get_detailed_h_score(state)
    best_state = list(state)
    best_h = current_h

    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_h = get_detailed_h_score(neighbor)
                if neighbor_h < best_h:
                    best_h = neighbor_h
                    best_state = neighbor
    return best_state, best_h

def print_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += "Q "
            else:
                line += ". "
        print(line)
    print(f"State representation: {state}\n")

def hill_climbing_with_summary(initial_state):
    """Run hill climbing, using restarts if stuck to guarantee finding the correct 0-conflict solution."""
    n = len(initial_state)
    current_state = list(initial_state)
    initial_h = get_detailed_h_score(current_state)

    print("==================== INITIAL STATE ====================")
    print(f"Initial State: {current_state}")
    print(f"h(N) Value (Conflicts): {initial_h}")
    print("=======================================================\n")

    # Run the search
    while True:
        current_h = get_detailed_h_score(current_state)
        if current_h == 0:
            break

        neighbor, neighbor_h = get_best_neighbor(current_state)
        if neighbor_h >= current_h:
            # If stuck in a local minimum, perform a random restart
            current_state = [random.randint(0, n - 1) for _ in range(n)]
            continue

        current_state = neighbor

    final_h = get_detailed_h_score(current_state)
    print("==================== FINAL STATE ====================")
    print(f"Final State: {current_state}")
    print(f"h(N) Value (Conflicts): {final_h}")
    print("=====================================================\n")

    print("Final Board Layout:")
    print_board(current_state)

# User Inputs
try:
    n = int(input("Enter number of queens (e.g., 4): "))
    print(f"Enter the initial state as a list of {n} row indices separated by spaces (e.g., 2 3 1 0):")
    state_input = input("Enter initial state: ")
    initial_state = [int(x) for x in state_input.split()]

    if len(initial_state) != n:
        raise ValueError("State length does not match the number of queens.")

    hill_climbing_with_summary(initial_state)
except ValueError as e:
    print(f"Error: {e}. Please run the cell again and provide valid inputs.")
```
